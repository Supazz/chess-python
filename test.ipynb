```python
%load_ext jupyter_black
import subprocess

def get_best_move(fen: str, depth: int = 4) -> str:
    
   
    try:
        # Call Engine.exe with FEN and depth as arguments
        result = subprocess.run(
            ["./Engine.exe", fen, str(depth)],
            capture_output=True,
            text=True,
            check=True
        )
        
        # Output from Engine.exe
        output = result.stdout.strip().splitlines()
        
    
        best_move = output[0].strip()
        eval = output[1].strip()
        return best_move, eval

        
    except subprocess.CalledProcessError as e:
        print("Error running engine:", e)
        return ""
    
```

```python
%load_ext jupyter_black
PIECE_VALUES = {
    chess.PAWN: 1,
    chess.ROOK: 5,
    chess.KNIGHT: 3,
    chess.BISHOP: 3,
    chess.QUEEN: 9,
    chess.KING: 0,
}


def evaluate_board(board: chess.Board):
    if board.is_game_over():
        color = board.outcome().winner
        if color == chess.WHITE:
            return float("inf")

        if color == chess.BLACK:
            return -float("inf")

    score = {chess.WHITE: 0, chess.BLACK: 0}

    for piece_type, value in PIECE_VALUES.items():
        score[chess.WHITE] += len(board.pieces(piece_type, chess.WHITE)) * value
        score[chess.BLACK] += len(board.pieces(piece_type, chess.BLACK)) * value
    return score[chess.WHITE] - score[chess.BLACK]


def minMax(board, depth, turn):
    best_move = None
    # Base case
    if depth == 0 or board.is_game_over():
        return evaluate_board(board), None
    # White wants to maximize evaulation
    if turn == chess.WHITE:
        max_eval = -float("inf")
        for move in board.legal_moves:
            # print(f"Legal Move {move}" )
            board.push(move)
            eval, _ = minMax(board, depth - 1, chess.BLACK)
            # print(eval)
            board.pop()
            if eval > max_eval:
                best_move = move
                # print(f"Eval Changed {best_move}")

                max_eval = eval
        # print(f"White Best move {best_move}, eval{max_eval}")
        return max_eval, best_move

    elif turn == chess.BLACK:
        min_eval = float("inf")
        for move in board.legal_moves:
            # print(f"Legal Move {move}" )

            board.push(move)
            eval, _ = minMax(board, depth - 1, chess.WHITE)
            board.pop()
            if eval < min_eval:
                best_move = move
                # print(f"Eval Changed {best_move}")
                min_eval = eval

        # print(f"Black Best move {best_move}, eval{min_eval}")
        return min_eval, best_move


def minMaxPruning(board, depth, alpha, beta, turn):
    best_move = None
    # Base case
    if depth == 0 or board.is_game_over():
        return evaluate_board(board), best_move
    # White wants to maximize evaulation
    if turn == chess.WHITE:
        max_eval = -float("inf")
        for move in board.legal_moves:
            # print(f"Legal Move {move}" )
            board.push(move)
            eval, _ = minMaxPruning(board, depth - 1, alpha, beta, chess.BLACK)
            # print(eval)
            board.pop()
            if eval > max_eval:
                best_move = move
                # print(f"Eval Changed {best_move}")
                max_eval = eval
            alpha = max(alpha, eval)
            if beta <= alpha:
                break

        # print(f"White Best move {best_move}, eval{max_eval}")
        return max_eval, best_move

    elif turn == chess.BLACK:
        min_eval = float("inf")
        for move in board.legal_moves:
            # print(f"Legal Move {move}" )
            board.push(move)
            eval, _ = minMaxPruning(board, depth - 1, alpha, beta, chess.WHITE)
            board.pop()
            if eval < min_eval:
                best_move = move
                # print(f"Eval Changed {best_move}")
                min_eval = eval
            beta = min(beta, eval)
            if beta <= alpha:
                break
        # print(f"Black Best move {best_move}, eval{min_eval}")
        return min_eval, best_move


def verify_legal_move(board: chess.Board, move: chess.Move) -> bool:
    if board.is_legal(move):
        return True


def get_user_move(board: chess.Board) -> chess.Move:
    # Verify Move Format
    while True:
        user_move_text = input()

        try:
            user_move = chess.Move.from_uci(user_move_text)
        except ValueError:
            # root = tk.Tk()
            # root.withdraw()
            print("Invalid Syntax!")
            continue

        # Verify Legal Move
        if not verify_legal_move(board, user_move):
            # root = tk.Tk()
            # root.withdraw()
            # messagebox.showwarning("Warning", "This move is illegal!")
            print("Not a legal move!")
        else:
            return user_move
```

```python
%load_ext jupyter_black


def play_game_python(depth):
    board = chess.Board()
    display(SVG(chess.svg.board(board=board, size=550)))

    while not board.is_game_over():
        userMoveText = input()

        userMove = chess.Move.from_uci(userMoveText)
        board.push(userMove)
        clear_output(wait=True)
        display(SVG(chess.svg.board(board=board, size=550)))

        _, computerMove = minMaxPruning(
            board, depth, -float("inf"), float("inf"), chess.BLACK
        )
        board.push(computerMove)

        clear_output(wait=True)
        display(SVG(chess.svg.board(board=board, size=550)))


def play_game_cpp(depth):
    board = chess.Board()
    clear_output(wait=True)
    display(SVG(chess.svg.board(board=board, size=550)))

    while not board.is_game_over():
        userMove = get_user_move(board)

        board.push(userMove)

        clear_output(wait=True)
        display(SVG(chess.svg.board(board=board, size=550)))

        computerMoveString, _ = get_best_move(board.fen(), depth)
        computerMove = chess.Move.from_uci(computerMoveString)
        board.push(computerMove)
        print(computerMove)

        clear_output(wait=True)
        display(SVG(chess.svg.board(board=board, size=550)))


def play_game_AI(depth, starting_fen=None):
    if starting_fen:
        board = chess.Board(starting_fen)
    else:
        board = chess.Board()
    display(SVG(chess.svg.board(board=board, size=550)))

    while not board.is_game_over():
        computerMoveString, _ = get_best_move(board.fen(), depth)
        computerMove = chess.Move.from_uci(computerMoveString)
        board.push(computerMove)

        clear_output(wait=True)
        display(SVG(chess.svg.board(board=board, size=550)))

        computerMoveString, _ = get_best_move(board.fen(), depth)
        computerMove = chess.Move.from_uci(computerMoveString)
        board.push(computerMove)

        clear_output(wait=True)
        display(SVG(chess.svg.board(board=board, size=550)))


def test_checkmate_white():
    board = chess.Board()
    board.push(chess.Move.from_uci("e2e4"))
    board.push(chess.Move.from_uci("g7g5"))
    board.push(chess.Move.from_uci("d2d4"))
    board.push(chess.Move.from_uci("f7f6"))
    board.push(chess.Move.from_uci("d1h5"))
    display(SVG(chess.svg.board(board=board, size=550)))
    print(evaluate_board(board))


def test_checkmate_black():
    board = chess.Board()
    board.push(chess.Move.from_uci("g2g4"))
    board.push(chess.Move.from_uci("e7e5"))
    board.push(chess.Move.from_uci("f2f4"))
    board.push(chess.Move.from_uci("d8h4"))
    display(SVG(chess.svg.board(board=board, size=550)))
    print(evaluate_board(board))
```
